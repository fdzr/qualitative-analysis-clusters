```python
import copy
import csv
from pathlib import Path

import pandas as pd
import numpy as np
```

```python
with open("gold-data-en.txt", "r") as f_in:
    gold_data = f_in.read().split("\n")

gold_data
```

```python
input_dataset = pd.read_json("inputs_en.data")
print(input_dataset.shape)
# full_dataset.rename(columns={"lemma": "word"}, inplace=True)
input_dataset.drop(columns=["start1", "end1", "start2", "end2"], inplace=True)
input_dataset.head()
```

```python
cont = 0
words_annotated = []

for d in Path("en_mcl/").rglob("*.scores"):
    word = d.stem.split(".")[1]
    if word in gold_data:
        words_annotated.append(pd.read_json(d))

annotations = pd.concat(words_annotated)
print(annotations.head())
print(annotations.shape)
```

```python
full_dataset = pd.merge(annotations, input_dataset)
print(full_dataset.shape)
full_dataset.rename(
    columns={"lemma": "word"},
    inplace=True,
)
full_dataset.drop(columns=["pos"], inplace=True)
full_dataset.head()
```

```python
full_dataset["score"] = full_dataset["score"].apply(
    lambda x: np.mean([float(v) for v in x])
)

full_dataset.head()
```

```python
gold_words_annotated = []
cont = 0

for d in Path("dwug_en/data").rglob("*judgments.csv"):
    cad = str(d)
    word = cad.split("/")[-2]

    if word in gold_data:
        data = pd.read_csv(d, sep="\t")
        data_cleaned = data[data["judgment"] != 0.0]

        gold_words_annotated.append(copy.deepcopy(data_cleaned))
```

```python
gold_data_annotated = pd.concat(gold_words_annotated)
gold_data_annotated.shape
gold_data_annotated.rename(columns={"lemma": "word"}, inplace=True)
gold_data_annotated.drop(columns=["annotator"], inplace=True)
```

```python
print(gold_data_annotated.shape)
gold_data_annotated.head()
```

```python
gold_data_annotated["_key"] = gold_data_annotated.apply(
    lambda row: tuple(sorted([row["identifier1"], row["identifier2"]])), axis=1
)
gold_data_annotated = gold_data_annotated.drop_duplicates(subset="_key").drop(
    columns="_key"
)
print(gold_data_annotated.shape)
gold_data_annotated.head()
```

```python
sentences_with_ids_gold_data = []
cont = 0

for d in Path("dwug_en/data").rglob("*uses.csv"):
    cad = str(d)
    word = cad.split("/")[-2]

    if word in gold_data:
        cont += 1
        data = pd.read_csv(
            d,
            sep="\t",
            engine="python",
            quoting=csv.QUOTE_NONE,
        )
        sentences_with_ids_gold_data.append(copy.deepcopy(data))

print(cont)
```

```python
sentences_gold_data = pd.concat(sentences_with_ids_gold_data)
print(sentences_gold_data.shape)
sentences_gold_data.drop(
    columns=[
        "indexes_target_token",
        "indexes_target_sentence",
        "indexes_target_token_tokenized",
        "indexes_target_sentence_tokenized",
        "context_lemmatized",
        "context_pos",
        "context_tokenized",
        "pos",
        "date",
        "description",
    ],
    inplace=True,
)
sentences_gold_data.rename(columns={"lemma": "word"}, inplace=True)
sentences_gold_data.head()
```

```python
result = pd.merge(
    gold_data_annotated,
    sentences_gold_data[["identifier", "context"]],
    left_on="identifier1",
    right_on="identifier",
    how="left",
)

result = result.rename(columns={"context": "sentence1"}).drop(columns="identifier")
result.head()
```

```python
result = pd.merge(
    result,
    sentences_gold_data[["identifier", "context"]],
    left_on="identifier2",
    right_on="identifier",
    how="left",
)
result = result.rename(columns={"context": "sentence2"}).drop(columns=["identifier", "comment"])
result.head()
```

```python
result.shape
```

```python
result_full_dataset = pd.merge(
    result,
    full_dataset,
    on=["word", "sentence1", "sentence2"],
    how="left",
)
result_full_dataset = result_full_dataset[
    [
        "identifier1",
        "sentence1",
        "identifier2",
        "sentence2",
        "word",
        "score",
        "judgment",
    ]
]
print(result_full_dataset.shape)
result_full_dataset.head()
```

```python
result_full_dataset["pair_key"] = result_full_dataset.apply(
    lambda row: tuple(sorted([row["identifier1"], row["identifier2"]])), axis=1
)
result_full_dataset = result_full_dataset.drop_duplicates(subset=["word", "sentence1", "sentence2"])
result_full_dataset = result_full_dataset.drop_duplicates(subset=["pair_key"])
result_full_dataset.drop(columns=["pair_key", "judgment"], inplace=True)
result_full_dataset.rename(columns={"score": "prediction"}, inplace=True)
print(result_full_dataset.shape)
result_full_dataset.head()
```

```python
result_full_dataset.dropna(inplace=True)
result_full_dataset.shape
```

```python
result_full_dataset.to_csv(
    "deepmistake_model_es.csv",
    index=False,
)
```
